```
# 補上四種資料型態的值
student_count = 60          # 整數 (int)
average_sleep_hours = 6.7   # 浮點數 (float)
course_name = "資料科學"    # 字串 (str)
has_missing_value = True    # 布林值 (bool)

assert isinstance(student_count, int), "student_count 應為整數。"
assert isinstance(average_sleep_hours, float), "average_sleep_hours 應為浮點數。"
assert isinstance(course_name, str), "course_name 應為字串。"
assert isinstance(has_missing_value, bool), "has_missing_value 應為布林值。"

print(type(student_count), type(average_sleep_hours))
print(type(course_name), type(has_missing_value))
```

```
# 串列：依順序儲存三個科系
departments = ["資訊管理", "企業管理", "金融管理"]
first_department = departments[0]  # 用索引取得第一個科系

assert first_department == departments[0], "請用串列索引取出第一個科系。"
print("全部科系：", departments)
print("第一個科系：", first_department)
print("科系數：", len(departments))

# 讓迴圈依序印出科系
loop_count = 0
for department in departments:
    print("科系：", department)
    loop_count = loop_count + 1

assert loop_count == len(departments), "請讓迴圈走訪全部科系。"

sample_student = {
    "student_id": 1,
    "department": "資訊管理",
    "sleep_hours": 7.0,
    "quiz_score": 82,
}
student_department = sample_student["department"]  # 用字典的鍵取得 department
student_score = sample_student["quiz_score"]        # 用字典的鍵取得 quiz_score

assert student_department == "資訊管理"
assert student_score == 82
print("學生科系：", student_department)
print("學生小考：", student_score)
```

```
# 一位假想學生的資料
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

base_score = 50
study_effect = 3 * study_hours           # 3 × 讀書時數
sleep_effect = 2 * sleep_hours           # 2 × 睡眠時數
screen_effect = -1.2 * screen_time_hours # -1.2 × 螢幕使用時數
attendance_effect = 0.15 * attendance_rate # 0.15 × 出席率

assert None not in (study_effect, sleep_effect, screen_effect, attendance_effect), "請先補完四項影響。"

sample_score = base_score
sample_score = sample_score + study_effect
sample_score = sample_score + sleep_effect
sample_score = sample_score + screen_effect
sample_score = sample_score + attendance_effect
print("模擬成績：", round(sample_score, 1))
```

```
# 補上三種分數範圍的回傳文字
def describe_score(score):
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

assert describe_score(82) == "表現良好", "檢查 80 分以上的分支。"
assert describe_score(70) == "達到基本標準", "檢查 60–79 分的分支。"
assert describe_score(55) == "需要加強", "檢查 60 分以下的分支。"

print(describe_score(82), describe_score(70), describe_score(55))
```

```
import numpy as np
import pandas as pd

random_seed = 42  # 填入課堂指定的整數種子 42

assert random_seed == 42, "請先使用種子 42，方便全班比較結果。"
rng = np.random.default_rng(random_seed)
check_rng = np.random.default_rng(random_seed)
print("兩次抽樣：", rng.random(), check_rng.random())

# 重設 rng，讓下一個 Mission 從相同起點抽樣
rng = np.random.default_rng(random_seed)
```

```
n = student_count
student_ids = np.arange(1, n + 1)  # 建立 1 到 60 的編號
chosen_departments = rng.choice(departments, size=n)
sleep_values = rng.normal(6.7, 1.1, size=n).round(1)

raw_screen_values = rng.normal(5.0, 1.5, size=n)
screen_values = raw_screen_values.clip(1, 10).round(1)  # 限制在 1 到 10，四捨五入

raw_study_values = rng.normal(2.5, 1.0, size=n)
study_values = raw_study_values.clip(0.5, 6).round(1)   # 限制在 0.5 到 6，四捨五入

attendance_values = rng.normal(88, 8, size=n)
attendance_values = attendance_values.clip(60, 100).round().astype(int)

assert student_ids is not None, "請先建立學生編號。"
assert screen_values is not None, "請先處理螢幕時間範圍。"
assert study_values is not None, "請先處理讀書時間範圍。"

df = pd.DataFrame({
    "student_id": student_ids,
    "department": chosen_departments,
    "sleep_hours": sleep_values,
    "screen_time_hours": screen_values,
    "study_hours": study_values,
    "attendance_rate": attendance_values,
})

display(df.head())
assert df.shape == (60, 6), "請檢查資料列數與欄數。"
```

```
# 沿用 Mission 4 的人為公式，逐項計算
study_part = 3 * df["study_hours"]
sleep_part = 2 * df["sleep_hours"]              # 2 × 睡眠時數欄位
screen_part = -1.2 * df["screen_time_hours"]    # -1.2 × 螢幕時間欄位
attendance_part = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, size=n)



raw_scores = 50 + study_part
raw_scores = raw_scores + sleep_part
raw_scores = raw_scores + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error
df["quiz_score"] = raw_scores.clip(0, 100).round().astype("Int64")

selected_columns = [ "student_id", "study_hours", "sleep_hours", "screen_time_hours", "attendance_rate", "quiz_score",]

display(df[selected_columns].head(60))
```
